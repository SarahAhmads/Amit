```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, root_mean_squared_error
import warnings
warnings.filterwarnings('ignore')
```

```python
Salary = pd.read_csv('Salary_Data.csv')
```

```python
plt.figure(figsize=(3,3))
sns.pairplot(Salary,x_vars=["YearsExperience"],y_vars=["Salary"],size=7,kind="scatter")
plt.xlabel("years")
plt.ylabel("salary")
plt.title("salary prediction")
plt.show()
```

```python
x = Salary.iloc[:,:-1]
y = Salary.iloc[:,1]
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=52)
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=15)
x_poly_train = poly.fit_transform(x_train)
x_poly_test = poly.transform(x_test)
poly_model = LinearRegression()
poly_model.fit(x_poly_train,y_train)
```

```python
y_poly_pred_train = poly_model.predict(x_poly_train)
y_poly_pred_test = poly_model.predict(x_poly_test)
```

```python
x_range = np.linspace(x.min(), x.max(), 100).reshape(-1, 1)
x_range_poly = poly.transform(x_range)
y_range_poly = poly_model.predict(x_range_poly)
```

```python
plt.scatter(x_train, y_train, color='blue', label='Training Data')
plt.scatter(x_test, y_test, color='green', label='Testing Data')
plt.plot(x_range, y_range_poly, color='red', label='Polynomial Fit')
plt.title('Salary vs Experience with Polynomial Fit')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

```python
mse = mean_squared_error(y_test, y_poly_pred_test)
rmse = root_mean_squared_error(y_test, y_poly_pred_test)
r2 = r2_score(y_test, y_poly_pred_test)
print(f'Mean Squared Error: {mse}')
print(f'Root Mean Squared Error: {rmse}')
print(f'R^2 Score: {r2}')
```
