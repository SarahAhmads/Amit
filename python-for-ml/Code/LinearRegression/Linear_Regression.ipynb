```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([2, 2.8, 3.6, 4.5])
```

```python
w=0
b=0
alpha=0.01
num_iters=20
```

```python
sse_value =[]

for i in range (num_iters):
    y_hat = w*x + b
    dw = (2/4) * np.sum((y_hat - y)*x)
    db = (2/4) * np.sum(y_hat - y)
    
    w -= alpha * dw
    b -= alpha * db
    
    sse = np.sum((y_hat - y) ** 2)
    sse_value.append(sse)

    if (i+1) % 20 == 0:
        print(f"Iteration {i+1}, SSE={sse}")
```

```python
print(f"oprimized param : slop = {w}, bias = {b}")
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(range(num_iters), sse_value, label='SSE')
plt.xlabel('Iterations')
plt.ylabel('SSE')
plt.title('SSE over Iterations')
plt.legend()
plt.subplot(1, 2, 2)
plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, w*x + b, color='red', label='Regression Line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.show()
```
