```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def sigmoid_derivative(output):
    return output * (1 - output)
```

```python
x1 = 0.05
x2 = 0.10

target1 = 0.01
target2 = 0.99

w1, w2, w3, w4 = 0.15, 0.20, 0.25, 0.30
w5, w6, w7, w8 = 0.40, 0.45, 0.50, 0.55

b1 = 0.35
b2 = 0.60

learning_rate = 0.5
```

```python
for epoch in range(10000):

    net_h1 = x1*w1 + x2*w2 + b1
    net_h2 = x1*w3 + x2*w4 + b1

    h1 = sigmoid(net_h1)
    h2 = sigmoid(net_h2)

    net_o1 = h1*w5 + h2*w6 + b2
    net_o2 = h1*w7 + h2*w8 + b2

    o1 = sigmoid(net_o1)
    o2 = sigmoid(net_o2)

    error = 0.5 * ((target1 - o1)**2 + (target2 - o2)**2)

    if epoch == 0:
        print("First iteration:")
        print("Output o1:", o1)
        print("Output o2:", o2)
        print("Error:", error)
        print("-" * 40)

    dE_o1 = (o1 - target1)
    dE_o2 = (o2 - target2)

    dO1_net = sigmoid_derivative(o1)
    dO2_net = sigmoid_derivative(o2)

    dw5 = dE_o1 * dO1_net * h1
    dw6 = dE_o1 * dO1_net * h2
    dw7 = dE_o2 * dO2_net * h1
    dw8 = dE_o2 * dO2_net * h2

    dH1 = (dE_o1 * dO1_net * w5 + dE_o2 * dO2_net * w7) * sigmoid_derivative(h1)
    dH2 = (dE_o1 * dO1_net * w6 + dE_o2 * dO2_net * w8) * sigmoid_derivative(h2)

    dw1 = dH1 * x1
    dw2 = dH1 * x2
    dw3 = dH2 * x1
    dw4 = dH2 * x2

    w1 -= learning_rate * dw1
    w2 -= learning_rate * dw2
    w3 -= learning_rate * dw3
    w4 -= learning_rate * dw4

    w5 -= learning_rate * dw5
    w6 -= learning_rate * dw6
    w7 -= learning_rate * dw7
    w8 -= learning_rate * dw8
```

```python
print("After 10,000 iterations:")
print("Final Output o1:", o1)
print("Final Output o2:", o2)
print("Final Error:", error)
```
