```python
import os
import random

def thanos_Project(folder_name, num_files):
    os.makedirs(folder_name)

    for i in range(num_files):
        file_path = os.path.join(folder_name, f"file_{i+1}.txt")
        with open(file_path, 'w') as f:
            f.write(f"This is a file")

    files = os.listdir(folder_name)
    files_to_delete = random.sample(files, (len(files) // 2))

    for file_name in files_to_delete:
        os.remove(os.path.join(folder_name, file_name))

folder_name = input("Enter folder name: ")
num_files = int(input("Enter number of files to create: "))
thanos_Project(folder_name, num_files)
print("Number of files remaining in the folder: ", len(os.listdir(folder_name)))
```
