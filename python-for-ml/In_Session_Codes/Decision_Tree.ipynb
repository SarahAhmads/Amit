```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

```python
Tennis = pd.read_csv(r'E:\Amit Repository\Amit\python-for-ml\In_Session_Codes\Datasets\PlayTennis.csv')
```

```python
from sklearn.preprocessing import OrdinalEncoder
outlook_encoder = OrdinalEncoder()
Tennis['Outlook'] = outlook_encoder.fit_transform(Tennis[['Outlook']])

temperature_encoder = OrdinalEncoder()
Tennis['Temperature'] = temperature_encoder.fit_transform(Tennis[['Temperature']])

humidity_encoder = OrdinalEncoder()
Tennis['Humidity'] = humidity_encoder.fit_transform(Tennis[['Humidity']])

windy_encoder = OrdinalEncoder()
Tennis['Wind'] = windy_encoder.fit_transform(Tennis[['Wind']])

play_encoder = OrdinalEncoder()
Tennis['Play Tennis'] = play_encoder.fit_transform(Tennis[['Play Tennis']])
```

```python
Tennis
```

```python
x = Tennis.drop('Play Tennis', axis=1)
y = Tennis['Play Tennis']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
Classifier = DecisionTreeClassifier(criterion='gini')
Classifier.fit(x_train, y_train)
y_pred = Classifier.predict(x_test)
```

```python
Classifier_1 = DecisionTreeClassifier(criterion='entropy')
Classifier_1.fit(x_train, y_train)
y_pred_1 = Classifier_1.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
Accuracy = accuracy_score(y_test, y_pred)
print('Accuracy of the model is :\n', Accuracy)
Report = classification_report(y_test, y_pred)
print('Classification Report with gini :\n', Report)
Conf_Matrix = confusion_matrix(y_test, y_pred)
print('Confusion Matrix :\n', Conf_Matrix)
```

```python
Accuracy = accuracy_score(y_test, y_pred_1)
print('Accuracy of the model is :\n', Accuracy)
Report = classification_report(y_test, y_pred_1)
print('Classification Report with entropy :\n', Report)
Conf_Matrix = confusion_matrix(y_test, y_pred_1)
print('Confusion Matrix :\n', Conf_Matrix)
```
