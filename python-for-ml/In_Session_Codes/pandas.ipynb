```python
data1 = [[1,444,'abc'],
        [2,555,'def'],
        [3,666,'ghi'],
        [4,np.nan,'jkl']
       ]
df = pd.DataFrame(data1, columns=['ID','std_id','Dept'])
null = df.isnull().sum()
pd.DataFrame(null).T
```

```python
def isNull(df):
    null = df.isnull().sum()
    return pd.DataFrame(null).T

data1 = [[1,444,'abc'],
        [2,555,'def'],
        [3,666,'ghi'],
        [4,np.nan,'jkl']
       ]
df = pd.DataFrame(data1, columns=['ID','std_id','Dept'])
isNull(df)
```
