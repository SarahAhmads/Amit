```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#This is the sample 2 dimensional data we are going to reduce
 
x = np.matrix([2.5,0.5,2.2, 1.9, 3.1, 2.3, 2, 1, 1.5, 1.1  ])
 
y = np.matrix([2.4,0.7,2.9,2.2,3.0,2.7,1.6,1.1,1.6,0.9])
original_data = np.r_[x,y]
```

```python
plt.scatter([x], [y])
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Original Data')
```

```python
from numpy import mean

mean_x = mean(x)
mean_y = mean(y)
x = x - mean_x
y = y - mean_y
```

```python
plt.scatter([x], [y])
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Original Data')
```

```python
data = np.r_[x,y]
covariance_matrix = np.matmul(data, data.T)/(10-1)
print(covariance_matrix)
```

```python
evals, evecs = np.linalg.eig(covariance_matrix)
print(evals)
print(evecs)
```

```python
evecs = np.flip(evecs, axis=1)
evals = np.flip(evals)
print(evals)
print(evecs)
```

```python
print(np.dot(evecs[:,0].T,evecs[:,1]))
```

```python
plt.plot([-evecs[0,1]*2, evecs[0,1]*2], [-evecs[1,1]*2, evecs[1,1]*2],'y')
plt.plot([-evecs[0,0]*2, evecs[0,0]*2], [-evecs[1,0]*2, evecs[1,0]*2],'r')
plt.scatter([x], [y])
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Centered Data with Eigen Vectors')
plt.axis('equal')
```
