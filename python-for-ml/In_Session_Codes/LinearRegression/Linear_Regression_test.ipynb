```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, root_mean_squared_error
import warnings
warnings.filterwarnings('ignore')
```

```python
Salary = pd.read_csv(r'E:\Amit Repository\Amit\python-for-ml\In_Session_Codes\Datasets\Salary_Data.csv')
Salary.head()
```

```python
plt.figure(figsize=(13,3))
sns.pairplot(Salary, x_vars=['YearsExperience'], y_vars='Salary', size=7, kind='scatter')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary Prediction')
plt.show()
```

```python
x = Salary.iloc[:, :-1].values
y = Salary.iloc[:, 1].values
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.8, random_state=10, shuffle=True)
```

```python
linear_model = LinearRegression()
linear_model.fit(x_train, y_train)
```

```python
plt.scatter(x_train, y_train, color='blue')
plt.plot(x_train, linear_model.predict(x_train), color='red')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary vs Experience')
plt.show()
```

```python
plt.scatter(x_test, y_test, color='blue')
plt.plot(x_test, linear_model.predict(x_test), color='red')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary vs Experience')
plt.show()
```

```python
#calculate the mean squared error and r2 score
y_pred = linear_model.predict(x_test)
mse = mean_squared_error(y_test, y_pred)
rms = root_mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rms)
print("R2 Score:", r2)
```
