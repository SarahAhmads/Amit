```python
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])
```

```python
Lin_Reg = LinearRegression()
Lin_Reg.fit(x, y)
y_pred_Lin = Lin_Reg.predict(x)
print(Lin_Reg.coef_)
print(Lin_Reg.intercept_)
```

```python
#Ridge
Lin_Ridge = Ridge(alpha=10.0)
Lin_Ridge.fit(x, y)
y_pred_Ridge = Lin_Ridge.predict(x)
print(Lin_Ridge.coef_)
print(Lin_Ridge.intercept_)
```

```python
#Lasso
Lin_Lasso = Lasso(alpha=10.0)
Lin_Lasso.fit(x, y)
y_pred_Lasso = Lin_Lasso.predict(x)
print(Lin_Lasso.coef_)
print(Lin_Lasso.intercept_)
```

```python
print("Linear Regression: ")
print("Mean Squared Error:", mean_squared_error(y, y_pred_Lin))
print("R2 Score:", r2_score(y, y_pred_Lin))
print("\nRidge Regression: ")
print("Mean Squared Error:", mean_squared_error(y, y_pred_Ridge))
print("R2 Score:", r2_score(y, y_pred_Ridge))
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred_Lin, color='red', label='Linear Regression')
plt.title('Linear Regression')
plt.legend()
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred_Ridge, color='red', label='Ridge Regression')
plt.title('Ridge Regression')
plt.legend()
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred_Lasso, color='red', label='Lasso Regression')
plt.title('Lasso Regression')
plt.legend()
```

```python
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred_Lin, color='red', label='Linear Regression')
plt.plot(x, y_pred_Ridge, color='green', linestyle='--', label='Ridge Regression')
plt.plot(x, y_pred_Lasso, color='orange', linestyle=':', label='Lasso Regression')
plt.title('Model Comparison')
plt.legend()
```
