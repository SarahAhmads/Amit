```python
#Task(2)_1: Email Checker
class EmailChecker:
    """
    This class extracts the username, domain name, and identify the domain type of an email address

    Attributes:
    email (str): The email address we want to check

    """
    def __init__(self, email):
        self.email = email

    def analyze_email(self):
        """
        This method analyzes the email to extract username, domain, and determine the domain type

        Returns:
        str: username, domain, and domain type
        
        """
        username = self.email.split("@")[0]
        domain = self.email.split("@")[1].split(".")
        domain_name = domain[0]

        if domain[-1] == "edu":
            domain_type = "Educational Domain"
        elif domain[-1] == "com":
            domain_type = "Commercial Domain"
        else:
            domain_type = "Other Domain"
        return (
            "The Username is: " + username + "\n"
            "The Domain is: " + domain_name + "\n"
            + domain_type
        )
    
email_address = input("Enter your email address: ")
email = EmailChecker(email_address)
print(email.analyze_email())
```

```python
#Task(2)_2: Message Decoder

class MessageDecoder:
    """
    This class decodes an encoded message

    Attributes:
        encoded_text (str): The encoded message we want to decoded
    """

    def __init__(self, encoded_text):
        self.encoded_text = encoded_text

    def extract_core(self):
        """
        This method extracts the core message without numbers and symbols

        Returns:
            tuple: The first and second words of the core message
        """
        core_part = ''
        letters = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz"
        for char in self.encoded_text:
            if char in letters or char == ' ': 
                core_part += char

        words = core_part.split()
        if len(words) >= 2:
            return words[0], words[1]
        else:
            return words[0], ''

    def shift_vowels(self, seconed_word):
        """
        This method replaces shifted vowels in the word to its orignal vowel
        """
        vowel_shift = {'A':'E', 'E':'I', 'I':'O', 'O':'U', 'U':'A', 'a':'e', 'e':'i', 'i':'o', 'o':'u', 'u':'a'}
        result = ''
        for char in seconed_word:
            if char in vowel_shift:
                result += vowel_shift[char] 
            else:
               result += char
        return result
    
    def decode(self):
        """
        This method reverse the first word and shift vowels in the second word.
        Returns: the decoded message.
        """
        first_word, second_word = self.extract_core()
        first_word_decoded = first_word[::-1]
        second_word_decoded = self.shift_vowels(second_word)
        return first_word_decoded + " " + second_word_decoded
    

encoded_message = input("Enter your encoded message: ")
decoder = MessageDecoder(encoded_message)
decoded_message = decoder.decode()
print("Decoded Message:", decoded_message)
```

```python
#Task(3): Add item to a tuple

class TupleModifier:
    """
    This class allows adding an item to a tuple
    
    Attributes:
        my_tuple (tuple): The original tuple to update
        item_to_add (any): The item to add to the tuple
    """

    def __init__(self, my_tuple, item_to_add):
        self.my_tuple = my_tuple  # Changed from self.original_tuple
        self.item_to_add = item_to_add

    def add_item(self):
        """
        This method adds the item to the tuple

        Returns:
            tuple: A new tuple with the new item added 
        """
        list_tuple = list(self.my_tuple)  # Changed from self.original_tuple
        list_tuple.append(self.item_to_add)
        updated_tuple = tuple(list_tuple)
        return updated_tuple

input_tuple = tuple(int(x.strip()) for x in input("Enter your tuple: ").split(',') if x.strip())
item_to_add = int(input("Enter an item to add to the tuple: "))
tuple_object = TupleModifier(input_tuple, item_to_add)
updated_tuple = tuple_object.add_item()
print(updated_tuple)
```

```python
#Task(3)_4: sum elements of a list

class SumList:
    """
    This class sums the elements of a list

    Attributes:
        my_list (list): The list of numbers to sum
    """
    def __init__(self, my_list):
        self.my_list = my_list

    def sum_elements(self):
        """
        This method sums the elements of the list

        Returns:
           sum (int): The sum of the list elements
        """
        sum = 0
        for num in self.my_list:
            sum += num
        return sum
    
input_list = [int(x.strip()) for x in input("Enter your list: ").split(',') if x.strip()]
sum_object = SumList(input_list)
total_sum = sum_object.sum_elements()
print("Sum of list elements:", total_sum)
```

```python
#Task(3)_5: Multipy items in a list

class MultiplyList:
    """
    This class multiplies the elements of a list

    Attributes:
        my_list (list): The list of numbers to multiply
    """
    def __init__(self, my_list):
        self.my_list = my_list

    def multiply_elements(self):
        """
        This method multiplies the elements of the list

        Returns:
           product (int): The product of the list elements
        """
        product = 1
        for num in self.my_list:
            product *= num
        return product
    
input_list = [int(x.strip()) for x in input("Enter your list: ").split(',') if x.strip()]
multiply_object = MultiplyList(input_list)
total_product = multiply_object.multiply_elements()
print("The multiplication of all list elements:", total_product)
```

```python
#Task(3)_6: Find the smallest number in a list

class FindSmallest:
    """
    This class finds the smallest number in a list

    Attributes:
        numbers (list): The list of numbers to search
    """

    def __init__(self, numbers):
        self.numbers = numbers

    def find_smallest(self):
        """
        This method finds the smallest number in the list

        Returns:
           smallest (int): The smallest number in the list
        """
        smallest = self.numbers[0]
        for num in self.numbers:
            if num < smallest:
                smallest = num
        return smallest

input_list = [int(x.strip()) for x in input("Enter your list: ").split(',') if x.strip()]
smallest_object = FindSmallest(input_list)
smallest_number = smallest_object.find_smallest()
print("The smallest number in the list is:", smallest_number)
```

```python
#Task(3)_8: Count numbers of string with length >= 2

class StringCounter:
    """
    This class counts the number of strings with length greater than or equal to 2

    Attributes:
        words (str): the list of words we want to count
    
    """
    def __init__(self, words):
        self.words = words
    
    def count_strings(self):
        """
        This method counts how many strings have length >= 2

        Returns:
            count (int): The count of strings with length >= 2
        """
        count = 0
        for word in self.words:
            if len(word) >= 2:
                count += 1
        return count
    
input_list = [x.strip() for x in input("Enter words: ").split(',') if x.strip()]
counter_object = StringCounter(input_list)
words_count = counter_object.count_strings()
print("Number of strings with length >= 2:", words_count)
```

```python
#Task(3)_9: Clone a list

class CloneList:
    """
    This class copies/clones a list

    Attributes:
        my_list (list): the list we want to copy
    """
    def __init__(self, my_list):
        self.original_list = my_list

    def clone_list(self):
        """
        This method creates a copy of the original list

        Returns:
            cloned_list (list): the new list with the same elements as the original
        """
        cloned_list = self.original_list[:]
        return cloned_list
    
input_list = [int(x.strip()) for x in input("Enter your list: ").split(',') if x.strip()]
clone_object = CloneList(input_list)
cloned_list = clone_object.clone_list()

print("Original List:", input_list)
print("Cloned List:  ", cloned_list)
```

```python
#Task(3)_10: Remove item from a set

class SetModifier:
    """
    This class removes an item from a set

    Attributes:
        my_set (set): The original set to update
    """
    def __init__(self, my_set):
        self.my_set = my_set

    def remove_item(self, item):
        """
        This method removes an item from the original set

        Args:
            item (any): The item to remove from the set
        """
        self.my_set.discard(item)

input_set = set(int(x.strip()) for x in input("Enter your set: ").split(',') if x.strip())
item_to_remove = int(input("Enter an item to remove from the set: "))
set_object = SetModifier(input_set)
updated_set = set_object.remove_item(item_to_remove)
print("Updated Set:", updated_set)
```

```python
#Task(3)_11: Check if a set is a subset of another set

class SubsetChecker:
    """
    This class checks if a set is a subset of another set

    Attributes:
        set1 (set): The first set
        set2 (set): The second set
    """
    def __init__(self, set1, set2):
        self.set1 = set1
        self.set2 = set2

    def is_subset(self):
        """
        This method checks if the first set is a subset of the second

        Returns:
            bool: True if the first set is a subset of the second, False otherwise
        """
        return self.set1.issubset(self.set2)

input_set1 = set(int(x.strip()) for x in input("Enter your first set: ").split(',') if x.strip())
input_set2 = set(int(x.strip()) for x in input("Enter your second set: ").split(',') if x.strip())
subset_checker = SubsetChecker(input_set1, input_set2)
is_subset_result = subset_checker.is_subset()

print("First Set:", input_set1)
print("Second Set:", input_set2)
print("Is the first set a subset of the second?", is_subset_result)
```

```python
#Task(3)_13: Find max and min values in a set

class FindMinMax:
    """
    This class finds the maximum and minimum values in a set

    Attributes:
        my_set (set): The original set
    """
    def __init__(self, my_set):
        self.my_set = my_set

    def find_min_max(self):
        """
        This method finds the maximum and minimum values in the original set
        """
        return max(self.my_set), min(self.my_set)

input_set = set(int(x.strip()) for x in input("Enter your set: ").split(',') if x.strip())
set_object = FindMinMax(input_set)
max_val = set_object.find_min_max()[0]
min_val = set_object.find_min_max()[1]
print("Maximum Value:", max_val)
print("Minimum Value:", min_val)
```

```python
#Task(3)_15: convert a tuple to a dictionary

class TupleConverter:
    """
    This class converts a tuple to a dictionary

    Attributes:
        my_tuple (tuple): The original tuple
    """
    def __init__(self, my_tuple):
        self.my_tuple = my_tuple

    def convert_to_dict(self):
        """
        This method converts the original tuple to a dictionary
        """
        result_dict = {}
        for i in range(0, len(self.my_tuple), 2):
            key = self.my_tuple[i]
            value = self.my_tuple[i+1]
            result_dict[key] = value
        return result_dict
    
input_tuple = tuple(int(x.strip()) for x in input("Enter your tuple: ").split(',') if x.strip())
tuple_object = TupleConverter(input_tuple)
dict_result = tuple_object.convert_to_dict()
print("Dictionary:", dict_result)
```

```python
#Task(3)_17: Reverse a tuple

class TupleReverser:
    """
    This class reverses a tuple

    Attributes:
        my_tuple (tuple): The original tuple
    """
    def __init__(self, my_tuple):
        self.my_tuple = my_tuple

    def reverse_tuple(self):
        """
        This method reverses the original tuple
        """
        return self.my_tuple[::-1]
    
input_tuple = tuple(int(x.strip()) for x in input("Enter your tuple: ").split(',') if x.strip())
tuple_object = TupleReverser(input_tuple)
reversed_tuple = tuple_object.reverse_tuple()
print("Reversed Tuple:", reversed_tuple)
```

```python
#Task(3)_19: Replace the last value of each tuple in a list with a given value
class TupleListModifier:
    """
    This class replaces the last value of each tuple in a list with a given value

    Attributes:
        tuple_list (list): The list of tuples to modify
        new_value (any): The new value to replace the last element of each tuple
    """
    def __init__(self, tuple_list, new_value):
        self.tuple_list = tuple_list
        self.new_value = new_value

    def replace_last_values(self):
        """
        This method replaces the last value of each tuple in the list with the new value

        Returns:
            modified_list (list): A new list of tuples with the last values replaced
        """
        modified_list = []
        for tup in self.tuple_list:
            if len(tup) > 0:
                modified_tuple = tup[:-1] + (self.new_value,)
                modified_list.append(modified_tuple)
            else:
                modified_list.append(tup)
        return modified_list
    
input_list_of_tuples = [tuple(int(x.strip()) for x in item.split(',')) for item in input("Enter your list of tuples: ").split(' ') if item.strip()]
new_value = int(input("Enter the new value to replace the last element of each tuple: "))
modifier_object = TupleListModifier(input_list_of_tuples, new_value)
modified_tuples = modifier_object.replace_last_values()
print("Modified List of Tuples:", modified_tuples)
```
